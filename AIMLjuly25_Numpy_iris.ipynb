A z-score is a statistical measurement that indicates how many standard deviations a data point is away from the mean of a dataset. It's used to determine whether a specific value is typical or unusual within a distribution. A positive z-score means the value is above the mean, while a negative z-score indicates it's below the mean.

**Purpose:**

Comparison: Z-scores allow for comparing data points from different datasets that might have different means and standard deviations.

Outlier Detection: Z-scores help identify outliers or unusual values within a dataset.

Probability Calculation: Z-scores can be used to determine the probability of a value occurring within a standard normal distribution.

Formula:
The z-score is calculated using the following formula:

z = (X - μ) / σ

Where:
z: is the z-score
X: is the individual data point
μ: is the population mean
σ: is the population standard deviation

Example:
If a student scores 85 on a test with a **mean** of 70 and a **standard deviation** of 10, the **z-score** would be:

z = (85 - 70) / 10 = 1.5

This means the student's score is 1.5 standard deviations above the mean.

#### Determinant and Inverse of iris.T @ iris

The 4×4 matrix (X^T × X) is square, so we can compute its determinant and inverse.

```python
# Determinant
det_XTX = np.linalg.det(XT_X)
print("Determinant of X.T @ X:", det_XTX)

# Inverse (only if determinant is non-zero)
if not np.isclose(det_XTX, 0):
    inv_XTX = np.linalg.inv(XT_X)
    print("Inverse of X.T @ X:\n", inv_XTX)
else:
    print("Matrix is singular and not invertible.")
```

#### Eigenvalues and Eigenvectors


```python
#  extract the eigenvalues and eigenvectors from the 4×4 matrix X.T @ X.
eigen_values, eigen_vectors = np.linalg.eig(XT_X)

print("Eigenvalues:\n", eigen_values)
print("Eigenvectors (first 2 columns):\n", eigen_vectors[:, :2])
```
